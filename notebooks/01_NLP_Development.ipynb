```python
import re

text = "Looking for a villa in Pune with swimming pool under 2 crore."

match = re.search(r"(\d+)\s*BHK", text, re.IGNORECASE)

if match:
    print("BHK:", match.group(1))
else:
    print("No BHK found")
```

```python
budget = re.search(r"(\d+)\s*(lakh|crore)", text, re.IGNORECASE)

if budget:
    print("Budget:", budget.group())
else:
    print("No budget found")
```

```python
property_types = [
    "apartment",
    "flat",
    "villa",
    "plot",
    "bungalow",
    "office",
    "shop"
]

found_property = None

for p in property_types:
    if p.lower() in text.lower():
        found_property = p
        break

print("Property Type:", found_property)
```

```python
locations = [
    "Baner",
    "Pune",
    "Mumbai",
    "Nashik",
    "Wakad",
    "Hinjewadi"
]

location = None

for city in locations:
    if city.lower() in text.lower():
        location = city
        break

print("Location:", location)
```

```python
amenities = [
    "gym",
    "parking",
    "garden",
    "lift",
    "security",
    "swimming pool"
]

found = []

for amenity in amenities:
    if amenity.lower() in text.lower():
        found.append(amenity)

print(found)
```

```python
result = {
    "bhk": match.group(1) if match else None,
    "budget": budget.group() if budget else None,
    "property_type": found_property,
    "location": location,
    "amenities": found
}

print(result)
```

```python
import re

def extract_customer_requirements(text):

    # ---------- Property Type ----------
    property_types = [
        "apartment",
        "flat",
        "villa",
        "plot",
        "bungalow",
        "office",
        "shop"
    ]

    found_property = None

    for p in property_types:
        if p.lower() in text.lower():
            found_property = p
            break


    # ---------- Location ----------
    locations = [
        "Baner",
        "Pune",
        "Mumbai",
        "Nashik",
        "Wakad",
        "Hinjewadi"
    ]

    found_location = None

    for city in locations:
        if city.lower() in text.lower():
            found_location = city
            break


    # ---------- Amenities ----------
    amenities = [
        "gym",
        "parking",
        "garden",
        "lift",
        "security",
        "swimming pool"
    ]

    found_amenities = []

    for amenity in amenities:
        if amenity.lower() in text.lower():
            found_amenities.append(amenity)


    # ---------- BHK ----------
    bhk_match = re.search(r"(\d+)\s*BHK", text, re.IGNORECASE)

    bhk = bhk_match.group(1) if bhk_match else None


    # ---------- Budget ----------
    budget_match = re.search(
        r"(\d+)\s*(lakh|crore)",
        text,
        re.IGNORECASE
    )

    budget = budget_match.group() if budget_match else None


    # ---------- Final Dictionary ----------
    result = {
        "BHK": bhk,
        "Property Type": found_property,
        "Location": found_location,
        "Budget": budget,
        "Amenities": found_amenities
    }

    return result
```

```python
text = "I need a 2 BHK apartment in Baner under 80 lakh with gym and parking."

result = extract_customer_requirements(text)

print(result)
```
